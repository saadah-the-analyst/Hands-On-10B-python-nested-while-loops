# Hands-On Exercise 1: E-Commerce Product Combination Generator

```python
# starting data
sizes = ["Small", "Medium", "Large"]
colors = ["Black", "White", "Blue"]

# create an index varaible i for sizes starting at 0
i = 0

# outer while loop for sizes
while i < len(sizes):

# create  reset index varaible j for colors of each size
    j = 0

# inner loop to cycle through colors
    while j < len(colors):
        print(f"{sizes[i]} - {colors[j]}")
        j += 1         # move to the next color index
    print()             # prints a blank line after each size group
    i += 1                   # move to the next size index

# Additional Challenge: calculate and display totasssl variations
total_varaiations = len(sizes) * len(colors)
print(f"Total Product Variations: {total_varaiations}")
```

# Hands-On Exercise 2: Banking Transaction Risk

```python
# starting data
customers = ["C001", "C002", "C003"]
transactions = [45000, 125000, 75000]
risk_threshold = 100000

# create index varaible i for customers starting at 0
i = 0

# outer while loop for customers 
while i < len(customers):
    j = 0          # reset transaction index for each customer 

# inner loop for transactions
    while j < len(transactions):
        if transactions[j] > risk_threshold:            # check if the transaction exceeds the threshold
            print(f"Customer: {customers}[i] | Transaction: {transactions}[j] | Status REVIEW")
        j += 1      # increment transaction indek
    i += 1          #increment customer index
```

### Additional Challenge

```python
# starting data
customers = ["C001", "C002", "C003"]
transactions = [45000, 125000, 75000]
risk_threshold = 100000

# initialize counter varaibles
i = 0
review_count = 0
total_comparisons = 0

# outer while loop for customers 
while i < len(customers):
    j = 0          # reset transaction index for each customer 

# inner loop for transactions
    while j < len(transactions):
        total_comparisons += 1          # count every comparison mode

# check if transactions exceeds risk threshold
        if transactions[j] > risk_threshold:
            print(f"Customer: {customers[i]} | Transaction: {transactions[j]} | Status REVIEW")
            review_count += 1
            
        j += 1      # increment transaction indek
    i += 1          #increment customer index

print(f"Transactions Requiring Review: {review_count}")
print(f"Total Comparisons Performed: {total_comparisons}")
```

# Hands-On Exercise 3: Hospital Appointment Scheduling Matrix

```python
# starting data
doctors = ["Dr. Ahmed", "Dr. Grace", "Dr. Michael"]
time_slots = ["9:00 AM", "11:00 AM", "2:00 PM", "4:00 PM"]

i = 0      # Doctor index

# Outer loop for doctors
while i < len(doctors):
    j = 0         # reset time slot index for each doctor

# inner loop for time slots
    while j < len(time_slots):
        print(f"{doctors[i]} - {time_slots[j]}")
        j += 1             # increment time slot index
    print()                # print a blank line to seperate each doctor's group
    i += 1                  # increment doctor index
```
